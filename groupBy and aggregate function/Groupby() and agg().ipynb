### Groupby() and agg()

```python
import pandas as pd
```

```python
df_cars= pd.read_csv('Car_sales.csv')
df_cars = df_cars.iloc[:,0:9].drop(columns=['Model','__year_resale_value'],axis=1)
df_cars
```

```python
# group by manufactuer and calculate the mean
df_cars.groupby('Manufacturer').agg('min')
```

```python
# find the minimum and maximum values of each column
df_cars.groupby('Vehicle_type').agg(['min','max'])
```

```python
# set a name or aggregated column. find the min engine size and max horsepower
df_cars.groupby('Vehicle_type').agg(min_engine_size=('Engine_size','min'),
                                  max_horse_power=('Horsepower','max'))
```

```python
#Calculate the sum of sales in thousand and mean of price in thousand
df_cars.groupby('Manufacturer').agg(sum_in_sale=('Sales_in_thousands','sum'),
                                   mean_price =('Price_in_thousands','mean'))
```
