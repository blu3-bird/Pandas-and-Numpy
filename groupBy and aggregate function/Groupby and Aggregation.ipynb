### Exercise 2: Basic agg()

Using the same orders DataFrame, calculate the total revenue (sum of amount), average 

order value (mean of amount), and total number of orders (count).

```python
orders
print('Total Revenue:', orders['amount'].agg('sum'))
print(f'Mean', {orders['amount'].agg('mean')})
print('Count', orders['order_id'].count())

results = orders.drop(columns=['customer','product','city']).agg(['sum','mean','count'])['amount']
print(results)
```

### Question 3 (Beginner) - Different aggregations per column

You have employee data. Calculate the sum and mean of 'salary', and the min and max of 

'age' in a single operation.

```python
employees = pd.DataFrame({
    'name': ['John', 'Sarah', 'Mike', 'Lisa', 'Tom'],
    'age': [28, 34, 45, 29, 38],
    'salary': [50000, 65000, 75000, 58000, 70000],
    'department': ['IT', 'HR', 'IT', 'Finance', 'HR']
})
employees
```

```python
employees.agg({
    'salary':['mean','sum'],
    'age':['min','max']
}).T
```

### Question 4 (Intermediate) - agg() with axis parameter

You have monthly sales data for 3 products. Calculate the total sales for each month (sum across products).



```python
sales = pd.DataFrame({
    'Product_A': [100, 150, 120, 180],
    'Product_B': [200, 180, 210, 190],
    'Product_C': [150, 160, 140, 170]
})
sales.index = ['January', 'February', 'March', 'April']
sales
```

```python

sales.agg({
    'January' : ['sum'],
    'February':['sum'],
    'March':['sum'],
    'April':['sum']
},axis=0)
```

### Question 5 (Beginner) - Basic groupby()

You have student exam scores. Group by 'subject' and calculate the average score for each subject.

```python
scores = pd.DataFrame({
    'student': ['Alice', 'Bob', 'Charlie', 'Alice', 'Bob', 'Charlie'],
    'subject': ['Math', 'Math', 'Math', 'Science', 'Science', 'Science'],
    'score': [85, 78, 92, 88, 82, 95]
})
scores
```

```python
scores.groupby('subject')['score'].agg('mean')
```

### Question 6 (Intermediate) - groupby() with get_group()

You have sales data from different stores. First, group by 'store', then extract and display only the data for 'Store B'

```python
sales_data = pd.DataFrame({
    'store': ['Store A', 'Store B', 'Store A', 'Store C', 'Store B', 'Store C', 'Store A'],
    'product': ['Widget', 'Gadget', 'Gadget', 'Widget', 'Widget', 'Gadget', 'Widget'],
    'sales': [100, 150, 120, 90, 200, 110, 130],
    'profit': [20, 45, 30, 15, 60, 25, 35]
})
sales_data
```

```python
sales_data.groupby('store').get_group('Store B')
```

### Question 7 (Intermediate) - groupby() with multiple aggregations

You have product inventory data. Group by 'category' and find both the minimum and maximum price for each category in one operation.

```python
inventory = pd.DataFrame({
    'product': ['Laptop', 'Mouse', 'Keyboard', 'Monitor', 'Webcam', 'Headphones'],
    'category': ['Electronics', 'Accessories', 'Accessories', 'Electronics', 'Accessories', 'Accessories'],
    'price': [1200, 25, 75, 400, 80, 150],
    'stock': [15, 100, 50, 20, 45, 60]
})
inventory
```

```python
inventory.groupby('category').agg(['min','max'])['price']
```

### Question 8 (Intermediate) - Named aggregations

You have restaurant orders. Group by 'meal_type' and calculate the total revenue (sum of price) and average order size (mean of quantity), giving clear names to these aggregations.


```python
restaurant = pd.DataFrame({
    'order_id': [1, 2, 3, 4, 5, 6],
    'meal_type': ['Breakfast', 'Lunch', 'Breakfast', 'Dinner', 'Lunch', 'Dinner'],
    'price': [12.50, 25.00, 10.00, 35.00, 22.00, 40.00],
    'quantity': [1, 2, 1, 3, 2, 2]
})
restaurant
```

```python
restaurant.groupby('meal_type').agg(
    total_revenue = ('price','sum'),
    order_size = ('quantity', 'mean')
)
```

### Question 9 (Intermediate) - Using apply() with lambda

You have monthly expenses in thousands. Group by 'category', calculate the sum, and then convert it to actual dollar amounts (multiply by 1000)

```python
expenses = pd.DataFrame({
    'category': ['Housing', 'Food', 'Housing', 'Transport', 'Food', 'Transport'],
    'month': ['Jan', 'Jan', 'Feb', 'Jan', 'Feb', 'Feb'],
    'amount_thousands': [2.5, 0.8, 2.5, 0.5, 0.9, 0.6]
})
expenses
```

```python
expenses.groupby('category')['amount_thousands'].sum().apply(lambda x:x*1000 )
```

### Question 10 (Advanced) - filter() with groupby

You have sales data for different salespersons. Filter and keep only those salespersons whose total sales exceed $500.

```python
sales_team = pd.DataFrame({
    'salesperson': ['Tom', 'Jerry', 'Tom', 'Spike', 'Jerry', 'Tom', 'Spike'],
    'product': ['A', 'B', 'C', 'A', 'A', 'B', 'C'],
    'sales': [150, 200, 180, 120, 250, 170, 90]
})

sales_team
```

```python
def q10(sales_team):
    return sales_team['sales'].sum() >= 500

sales_team.groupby('salesperson').filter(q10)
```
