### Groupby()

```python
import pandas as pd
```

```python
df_cars = pd.read_csv('Car_sales.csv')
df_cars = df_cars.iloc[:,0:9].drop(columns=['Model','__year_resale_value'],axis=1)
df_cars
```

```python
#Group by vehicle type and calculate their max.
df_cars.groupby('Vehicle_type').max()
```

```python
#set different index instead of col name
df_cars.groupby('Vehicle_type',as_index=False).min()
```

```python
#Attributes

#print(df_cars.groupby('Manufacturer').groups)
df_cars.groupby('Manufacturer').groups.keys
```

```python
# get a specific grpup within the manufacturer col
df_cars.groupby('Manufacturer').get_group('Audi')
```

```python
#group by Manufacturer and calculate their sum
df_cars.groupby('Manufacturer')['Sales_in_thousands'].sum().head()
```

```python
#find the null values in columns
df_cars[df_cars['Engine_size'].isnull()]
```

```python
#group by engine size and count values without dropping the Nan values
df_cars.groupby('Engine_size',dropna=False).count()
```
