### GroupBy and aggregate function

```python
import pandas as pd
```

```python
# reading the csv file
df_cars = pd.read_csv('Car_sales.csv')
df_cars = df_cars[['Manufacturer','Sales_in_thousands','Vehicle_type','Price_in_thousands','Engine_size','Horsepower','Fuel_capacity']]
df_cars.head()
```

## agg() function

```python
# Calculate the sum of the columns.
df_cars.agg('sum')
```

```python
# To calculate the mean of the dateset
df_cars.agg('mean') #Getting the typeError, as there are some string elements and columns in the dataset.  # new pandas version
```

```python
# To calculate the mean and count variables
df_cars.apply(['max','sum']) # as mean function wont work here. 
```

```python
# Different aggregates per columns: calculate th sum and mean in 'Sales in thousands' and sum & mean of 'Price_in_thousands'.

df_cars.agg({'Sales_in_thousands':['sum','max'],
             'Price_in_thousands':['min','sum']})  

# here we null values for price_in_thousands cause we never called the functions for those columns.
```

### Aggregate over the columns

```python
df_cars[['Sales_in_thousands','Price_in_thousands']].agg(['sum','mean'], axis=1)
```

```python
# Aggregate different functions and rename the index of the resulting dataframe
df_cars.agg(x=('Price_in_thousands','sum'), y =('Sales_in_thousands','mean'))
```

### The split-apply-combine strategy 

The Strategy helps us to group values in a dataframe. so we build categories and we can apply some aggre function. so, we can the values of these categories.

```python
# split data into separate groups
df_cars['Vehicle_type'].value_counts() # we gonna split dataframe into the two categories.
```

```python
# filtering the data
passenger_data = df_cars['Vehicle_type'] == 'Passenger'
car_data = df_cars['Vehicle_type'] == 'Car'
passenger_data
```

### apply an operation

```python
# filtered dataframe
df_cars[passenger_data] # we will get the other filtered dataframe by using car_data.
```

```python
# applying an operation 
cars_average = df_cars[car_data]['Sales_in_thousands'].agg('mean')
passenger_average = df_cars[passenger_data]['Sales_in_thousands'].agg('mean')
```

```python
#Combining the results
pd.DataFrame({'Vehicle_type':['Car','Passenger'],
              'Sales_in_thousands':[cars_average,passenger_average]})
```
