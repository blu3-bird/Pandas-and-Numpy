```python
import pandas as pd
```

```python


# Create a sample dataset
data = {
    'Store': ['North', 'North', 'North', 'South', 'South', 'South', 'East', 'East', 'West', 'West'],
    'Product': ['Apple', 'Banana', 'Orange', 'Apple', 'Banana', 'Orange', 'Apple', 'Banana', 'Apple', 'Orange'],
    'Category': ['Fruit', 'Fruit', 'Fruit', 'Fruit', 'Fruit', 'Fruit', 'Fruit', 'Fruit', 'Fruit', 'Fruit'],
    'Units_Sold': [150, 200, 80, 120, 180, 95, 200, 150, 90, 110],
    'Price_Per_Unit': [1.5, 0.75, 2.0, 1.6, 0.80, 1.9, 1.4, 0.70, 1.55, 2.1],
    'Revenue': [225, 150, 160, 192, 144, 180.5, 280, 105, 139.5, 231]
}

df = pd.DataFrame(data)
print(df)
```

#### Exercise Set 1: Basic Aggregation with agg()
Exercise 1.1 - Single Aggregation
Task: Calculate the sum of all values in the Units_Sold column using agg().

```python
df['Units_Sold'].agg('sum')
```

Exercise 1.2 - Mean Calculation
Task: Calculate the mean of all numeric columns using agg().

```python
df.iloc[:,-3:].agg('mean')
```

Exercise 1.3 - Multiple Aggregations
Task: Calculate both the mean and count for all columns using agg().

```python
df.iloc[:,-3:].agg(['count','mean'])
```

Exercise 1.4 - Different Aggregations per Column
Task: Calculate:

sum and max for Units_Sold

mean and min for Revenue

```python
df.agg({
    'Units_Sold':['sum','mean'],
    'Revenue':['mean','min']
})
```

Exercise 1.5 - Row-wise Aggregation

Task: Calculate the sum of Units_Sold and Revenue for each row (across columns).

```python
df[['Units_Sold','Revenue']].agg('sum',axis=1)
```

### Exercise Set 2: GroupBy Basics

Exercise 2.1 - Simple GroupBy

Task: Group the data by Store and calculate the mean of all numeric columns.

```python
df
```

```python
df.iloc[:,-3:].groupby(df['Store']).agg('mean')
```

Exercise 2.2 - GroupBy with Sum

Task: Group by Product and calculate the total (sum) of Units_Sold for each product.

```python
df.groupby('Product')['Units_Sold'].agg('sum')
```

Exercise 2.3 - Accessing Groups

Task: Create a groupby object, print all group keys, and get data for 'North' store only.

```python
df.groupby('Store').get_group('North')
```

Exercise 2.4 - GroupBy Count

Task: Group by Store and count how many products each store has.

```python
df.groupby('Store').count()['Product']
```

Exercise 2.5 - Unique Values

Task: Find how many unique values exist in each column using nunique().

```python
df.nunique()
```

### 📝 Exercise Set 3: GroupBy with agg()

Exercise 3.1 - Multiple Aggregations per Group

Task: Group by Store and find the min and max values for Units_Sold and Revenue.

```python
df.groupby('Store').agg(['min','max'])[['Units_Sold','Revenue']]
```

Exercise 3.2 - Named Aggregations

Task: Group by Product and calculate total units sold (name it total_units) and average revenue (name it avg_revenue).

```python
df.groupby('Product').agg(total_units=('Units_Sold','sum'),avg_revenue=('Revenue','mean'))
```

Exercise 3.3 - Complex Aggregation

Task: Group by Store and calculate sum of Units_Sold, mean of Price_Per_Unit, and max of Revenue.

```python
df.groupby('Store').agg({
    'Units_Sold':'sum',
    'Price_Per_Unit':'mean',
    'Revenue':'max'
})
```

### 📝 Exercise Set 4: Lambda Functions with GroupBy

Exercise 4.1 - Simple Lambda

Task: Group by Store, sum the Revenue, then multiply all values by 100 (convert to cents).

```python
df.groupby('Store')['Revenue'].sum().apply(lambda x: x*100 )
```

Exercise 4.2 - Subtract Group Mean

Task: Group by Product and subtract the mean Units_Sold from each value within the group.

```python
df.groupby('Product')['Units_Sold'].apply(lambda x: x-x.mean())
```

Exercise 4.3 - Custom Calculation (Range)

Task: Group by Store and calculate the range (max - min) of Revenue for each store.

```python
df.groupby('Store')['Revenue'].apply(lambda x: x.max() - x.min())
```

### 📝 Exercise Set 5: Challenge Exercises

Task: Find which Store has the highest average Revenue

```python
df.groupby('Store')['Revenue'].mean().idxmax()
```

Challenge 5.2 - Multiple Group Keys

Task: Group by both Store AND Product, then calculate the total Units_Sold.

```python
df.groupby(['Store','Product'])['Units_Sold'].sum()
```

Challenge 5.3 - Calculate Percentages

Task: Calculate what percentage of total revenue each store contributes.

```python
total_revenue = df['Revenue'].sum()
#print(total_revenue)

df.groupby('Store')['Revenue'].sum().apply(lambda x: (x/total_revenue*100))
```
