### Interactive Visualizatio with pandas

```python
import pandas as pd
import plotly.express as px
```

```python
# reading the csv file
df_population_raw = pd.read_csv('population_total.csv')
# dropping null values
df_population_raw.dropna(inplace=True)
# making a pivot table
df_pivot = df_population_raw.pivot(index='year', columns='country',
                                        values='population')
# selecting some countries
df_pivot = df_pivot[['United States', 'India', 'China', 
                          'Indonesia', 'Brazil']]
# showing pivot table
df_pivot
```

```python
# creating line plot with px method
px.line(df_pivot, title='Population over the year in some countries',
        labels={'value':'Population','index':'Years'})
```

```python
# to comparing two or more countries only 'y' parameter
px.line(df_pivot,title='Comparing the population between India nd China',
        labels={'value':'Population','year':'Years'},
        y=['China','India'])
```

### Bar plots


```python
#single bar plots
df_bar = df_pivot[df_pivot.index.isin([2016])]
df_bar_main = df_bar.T
df_bar
```

```python
px.bar(df_bar,labels={'value':'Population','country':'Country'},barmode='group')
```


### Barplots grouped by n variables

```python
#creating dataframe with some selected years
df_bar_n = df_pivot[df_pivot.index.isin([2005,2010,2015,2020])]
df_bar_n_t = df_bar_n.T
df_bar_n
```

```python
px.bar(df_bar_n,barmode='group',orientation='v')
```

### Pie chart

### Box plot

```python
df_box = df_pivot[['India','China','Brazil']]
px.box(df_pivot, x=['India','China'], labels={'variable':'Country','value':'Population'})
```

### Area plot

```python
px.area(df_pivot,labels={'value':'Population','year':'Years'})
```
