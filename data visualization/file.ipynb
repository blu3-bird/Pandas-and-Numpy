### Showing and saving the plots

```python
import pandas as pd
```

```python
import matplotlib.pyplot as plt
```

```python
df_raw = pd.read_csv('population_total.csv')
```

```python
df_raw.dropna(inplace=True)
```

```python
df_population = df_raw.pivot(index='year',columns='country',values='population')
```

```python
#selcting only few columns
df_main = df_population[['United States','India','China','Spain','Brazil']]
df_main
```

```python
#making the line plot
df_main.plot(kind='line' ,xlabel='Year', ylabel='Population', title='Population over the years in some countries')

#Showing the plot (without the outline code in output)
plt.show()

# saving the plots locally in device
plt.savefig('test.png')
```
