### Question 9 (Intermediate) - value_counts()

You have customer feedback data. Count how many times each rating appears, show the results in descending order with percentages.

```python
feedback = pd.DataFrame({
    'customer': ['John', 'Sarah', 'Mike', 'Lisa', 'Tom', 'Emma', 'David', 'Anna', 'Chris', 'Julie'],
    'rating': [5, 4, 5, 3, 5, 4, 5, 2, 4, 5]
})
feedback
```

```python
print('Rating count:')
rating_count = feedback.value_counts('rating')
print(rating_count)

print('Rating Percentage:')

rating_percentage = feedback.value_counts('rating',normalize=True) * 100

print(rating_percentage)

print('Final Results:')
final_result = pd.DataFrame({
    'count' : rating_count,\
    'percentage': rating_percentage
})
print(final_result)
```


### Question 4 (Intermediate) - apply() with custom function

You have temperature data in Celsius. Create a custom function to convert to Fahrenheit and apply it to the temperature column. Formula: F = (C × 9/5) + 32

```python
weather = pd.DataFrame({
    'city': ['New York', 'London', 'Tokyo', 'Sydney'],
    'temp_celsius': [20, 15, 25, 18]
})
weather
```

```python
weather['temp_f'] = weather['temp_celsius'].apply(lambda x : (x * (9/5)) + 32 )
weather
```

### Apply to entire dataframe (row-wise)

```python
def classify_temp(row):
    if row['temp_celsius'] > 20:
        return 'Hot'
    else:
        return 'Cold'
    
weather['Classification'] = weather.apply(classify_temp, axis=1)
weather
```

### Question 6 (Intermediate) - String methods with .str

You have customer data with messy email addresses (mixed case, spaces). Clean them by converting to lowercase and removing spaces.

```python
customers = pd.DataFrame({
    'name': ['Alice', 'Bob', 'Charlie', 'Diana'],
    'email': ['ALICE@Email.COM', ' bob@email.com ', 'Charlie@EMAIL.com', 'DIANA@email.com  ']
})

print("Original Data:")
print(customers)

print('\n\nIssues in emails:')

for i , email in enumerate(customers['email']):
    print(f'Row{i}:      " {email} "     Length: {len(email)}')

print("\nCleaned Data:")
customers['Cleaned_email'] = customers['email'].str.strip().str.lower()
print(customers)

print('\n\nCleaned DataSet:')
customers['Domain'] = customers['Cleaned_email'].str.split('@').str[1]
print(customers)
```

### Question 8 (Intermediate) - concat() to combine DataFrames

You have sales data from two different quarters stored in separate DataFrames. Combine them vertically (stack them) into one DataFrame.


```python
q1_sales = pd.DataFrame({
    'month': ['Jan', 'Feb', 'Mar'],
    'sales': [15000, 18000, 16000],
    'expenses': [8000, 9000, 8500]
})

q2_sales = pd.DataFrame({
    'month': ['Apr', 'May', 'Jun'],
    'sales': [19000, 21000, 20000],
    'expenses': [9500, 10000, 9800]
})

print("First DataFrame")
print(q1_sales)

print('\nSecond DataFrame')
print(q2_sales)


df = pd.concat([q1_sales, q2_sales])
df.reset_index(drop=True,inplace=True)

print('\n\n Combined DataFrame with new index:')
print(df)
```

### Question 9 (Intermediate) - merge() to join DataFrames

You have customer information in one DataFrame and their orders in another. Merge them to see customer names alongside their orders

```python
customers = pd.DataFrame({
    'customer_id': [1, 2, 3, 4],
    'name': ['Alice', 'Bob', 'Charlie', 'Diana'],
    'city': ['NYC', 'LA', 'Chicago', 'Boston']
})

orders = pd.DataFrame({
    'order_id': [101, 102, 103, 104, 105],
    'customer_id': [1, 2, 1, 3, 2],
    'product': ['Laptop', 'Mouse', 'Monitor', 'Keyboard', 'Webcam'],
    'amount': [1200, 25, 400, 75, 80]
})

print('Customer DataFrame:\n')
print(customers)

print('\n Customer Orders DataFrame:\n')
print(orders)

merge_df = pd.merge(customers,orders, on='customer_id', how='left')

print('\n\nMerged DataSet of both dataframes\n')
print(merge_df)
print('\n\n')


print('\nfinal dataset:\n')
summary_df = (
    merge_df.groupby('name')
    .agg({'product': 'count', 'amount': 'sum'})
    .rename(columns={'product': 'count', 'amount': 'spent'})
    .reset_index()
)
print(summary_df)



```

### Question 10 (Intermediate) - pivot_table() for reshaping data

You have sales data in long format (one row per transaction). Create a pivot table showing total sales for each salesperson by region.

```python
sales = pd.DataFrame({
    'salesperson': ['Tom', 'Jerry', 'Tom', 'Spike', 'Jerry', 'Tom', 'Spike', 'Jerry'],
    'region': ['North', 'South', 'South', 'North', 'North', 'North', 'South', 'South'],
    'product': ['A', 'B', 'A', 'C', 'C', 'B', 'A', 'A'],
    'sales': [150, 200, 180, 120, 250, 170, 90, 220]
})


sales_pivot = pd.pivot_table(
    sales,
    values='sales',
    columns='region',
    index='salesperson',
    aggfunc='sum',
    fill_value=0,
    margins=True # add grand total
)


print('Sales Transaction Data:\n', sales)
print('\nSales Pivot Table:\n',sales_pivot)
```
