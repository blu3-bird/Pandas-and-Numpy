### Question 1 (Beginner) - loc[] for label-based selection

You have a DataFrame of students with their test scores. Use loc[] to select the data for students 'Alice' and 'Charlie', showing only their 'math_score' and 'science_score' columns.

```python
import pandas as pd

students = pd.DataFrame({
    'name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve'],
    'math_score': [85, 78, 92, 88, 95],
    'science_score': [90, 82, 88, 91, 89],
    'english_score': [88, 85, 90, 87, 91]
})
students.set_index('name', inplace=True)
students
```

```python
students.loc[['Alice','Charlie'],['math_score','english_score']]
```

### Question 3 (Beginner) - sort_values()

You have employee data. Sort the employees by salary in descending order (highest to lowest).

```python
employees = pd.DataFrame({
    'name': ['John', 'Sarah', 'Mike', 'Lisa', 'Tom'],
    'department': ['IT', 'HR', 'IT', 'Finance', 'HR'],
    'salary': [65000, 70000, 58000, 75000, 62000],
    'age': [28, 34, 45, 29, 38]
})
employees
```

### Question 6 (Intermediate) - Multiple conditions with boolean filtering

From the products DataFrame, find all products that are:

    In the 'Accessories' category AND
    
    Priced between $50 and $200 (inclusive) AND
    
    Currently in stock


```python
products = pd.DataFrame({
    'product': ['Laptop', 'Mouse', 'Keyboard', 'Monitor', 'USB Cable', 'Headphones', 'Webcam'],
    'price': [1200, 25, 75, 400, 15, 150, 80],
    'category': ['Electronics', 'Accessories', 'Accessories', 'Electronics', 'Accessories', 'Accessories', 'Accessories'],
    'in_stock': [True, True, False, True, True, True, True]
})

products
```

```python
products[
    (products['category'] == 'Accessories') &
    (products['price'] > 50) &
    (products['price'] < 200) &
    (products['in_stock'] == True)
]
```

### Question 7 (Beginner) - fillna() to handle missing data

You have temperature readings with some missing values (NaN). Fill the missing values with the average temperature.

### Question 9 (Intermediate) - value_counts()

You have customer feedback data. Count how many times each rating appears, show the results in descending order with percentages.

```python
feedback = pd.DataFrame({
    'customer': ['John', 'Sarah', 'Mike', 'Lisa', 'Tom', 'Emma', 'David', 'Anna', 'Chris', 'Julie'],
    'rating': [5, 4, 5, 3, 5, 4, 5, 2, 4, 5]
})
feedback
```

```python
print('Rating count:')
rating_count = feedback.value_counts('rating')
print(rating_count)

print('Rating Percentage:')

rating_percentage = feedback.value_counts('rating',normalize=True) * 100

print(rating_percentage)

print('Final Results:')
final_result = pd.DataFrame({
    'count' : rating_count,\
    'percentage': rating_percentage
})
print(final_result)
```

### Question 10 (Intermediate) - drop_duplicates()

You have customer transaction data with duplicate entries. Remove duplicate rows, keeping only the first occurrence of each unique customer.

```python
transactions = pd.DataFrame({
    'customer': ['Alice', 'Bob', 'Alice', 'Charlie', 'Bob', 'Diana', 'Alice'],
    'product': ['Laptop', 'Mouse', 'Keyboard', 'Monitor', 'Keyboard', 'Laptop', 'Mouse'],
    'amount': [1200, 25, 75, 400, 75, 1200, 30]
})
transactions
```

```python
transactions.drop_duplicates('customer', keep='first')
```
